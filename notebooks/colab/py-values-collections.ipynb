# Python basics: values, strings, lists and dictionaries

Python programs work with a few built-in types. Numbers, strings, booleans and None hold single values; lists, tuples, dictionaries and sets hold collections. Knowing which are mutable, how indexing and slicing count, and that assignment shares rather than copies objects prevents most beginner bugs.

**Goal:** predict the result of each line, then run it.

**Part A: numbers and strings.**

````python
print(7 / 2, 7 // 2, 7 % 2)           # 3.5 3 1
name = "  asha  "
print(name.strip().title())           # Asha
print(f"{name.strip()} has {3} orders") # asha has 3 orders
print(0.1 + 0.2 == 0.3, round(0.1 + 0.2, 2) == 0.3)   # False True
````

**Part B: lists and slicing.**

````python
names = ["Asha", "Ben", "Chen", "Dee"]
print(names[0], names[-1], names[1:3])   # Asha Dee ['Ben', 'Chen']
````

**Part C: sharing versus copying.**

````python
a = [1, 2]
b = a
c = a.copy()
b.append(3)
print(a, c)                              # [1, 2, 3] [1, 2]
````

**Part D: dictionaries and sets.**

````python
row = {"id": 7, "city": "Leeds"}
print(row.get("country", "unknown"))     # unknown
seen = {1, 2, 3}
print(2 in seen, 9 in seen)              # True False
````

**Part E: the scenario.** Predict the total and the number of countries, then run:

````python
orders = [{"customer": "asha", "country": "UK", "amount": 120},
          {"customer": "ben", "country": "FR", "amount": 80},
          {"customer": "chen", "country": "UK", "amount": 50}]
total = 0
countries = set()
for order in orders:
    print(order["customer"].upper())
    total += order["amount"]
    countries.add(order["country"])
print(total, countries, len(countries))
````
