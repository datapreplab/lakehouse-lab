# Python: loops, conditions and functions

if/elif/else chooses a branch, for loops visit each item and while loops repeat until a condition fails. Functions package logic with parameters and a return value; a function without return gives None. Comprehensions, enumerate and zip make common loops shorter.

**Goal:** predict, run and fix small pieces of control flow.

**Part A: range and comprehensions.**

````python
print(list(range(0, 5, 2)))                       # [0, 2, 4]
print([i * i for i in range(4) if i % 2 == 1])    # [1, 9]
````

**Part B: enumerate and zip.**

````python
for i, name in enumerate(["Asha", "Ben"], start=1):
    print(i, name)
print(dict(zip(["a", "b"], [1, 2])))              # {'a': 1, 'b': 2}
````

**Part C: the missing return.**

````python
def add_tax(x):
    x * 1.2
print(add_tax(10))                                # None
````

Fix it with `return x * 1.2` and rerun.

**Part D: the mutable default.**

````python
def add(item, bucket=[]):
    bucket.append(item)
    return bucket
print(add(1), add(2))                             # [1, 2] [1, 2]
````

Fix it with `bucket=None` and `bucket = [] if bucket is None else bucket`, then rerun: `[1] [2]`.

**Part E: the scenario.** Predict the result, then run:

````python
def total_processed(sizes, limit=1000):
    total = 0
    for size in sizes:
        if size == 0:
            continue
        if size > limit:
            break
        total += size
    return total

print(total_processed([10, 0, 250, 1500, 40]))
````

Part B -> 1 Asha / 2 Ben ; {'a': 1, 'b': 2}
