# Tables and rows: CREATE, INSERT, UPDATE and DELETE

Before you can query data you need somewhere to keep it. A table has named, typed columns and lives at catalog.schema.table. CREATE TABLE defines it, INSERT adds rows, SELECT reads them, UPDATE changes them and DELETE removes them (CRUD). UPDATE and DELETE without WHERE change every row.

**How to run:** attach **Serverless** compute (top right), then run the cells from top to bottom. Each part asks you to predict the result first; the expected results are listed at the end.

From the Learning Hub lesson: [Tables and rows: CREATE, INSERT, UPDATE and DELETE](https://www.datapreplab.com/hub#topic/sql-tables-crud) · Independent study resource, not affiliated with or endorsed by Databricks.

**Goal:** create a table, then insert, read, update and delete rows, predicting each result first.

**Part A: create the table and look at its columns.**

````python
%sql
CREATE OR REPLACE TABLE workspace.default.customers (
  customer_id INT,
  name        STRING,
  city        STRING,
  joined      DATE
);
````

````python
%sql
DESCRIBE TABLE workspace.default.customers;
````

**Part B: create rows (INSERT) and read them (SELECT).**

````python
%sql
INSERT INTO workspace.default.customers VALUES
  (1, 'Asha', 'Leeds', DATE'2026-01-10'),
  (2, 'Ben',  'Leeds', DATE'2026-02-03'),
  (3, 'Chen', 'York',  DATE'2026-02-20'),
  (4, 'Dee',  'Hull',  DATE'2026-03-05');
````

````python
%sql
SELECT * FROM workspace.default.customers ORDER BY customer_id;
````

**Part C: update one row.** Preview which rows the condition matches, then change them:

````python
%sql
SELECT * FROM workspace.default.customers WHERE customer_id = 2;
````

````python
%sql
UPDATE workspace.default.customers SET city = 'York' WHERE customer_id = 2;
````

````python
%sql
SELECT * FROM workspace.default.customers WHERE customer_id = 2;
````

**Part D: delete one row.**

````python
%sql
DELETE FROM workspace.default.customers WHERE customer_id = 4;
````

````python
%sql
SELECT * FROM workspace.default.customers ORDER BY customer_id;
````

**Part E: see every change.** Each statement was saved as a new version of the table:

````python
%sql
DESCRIBE HISTORY workspace.default.customers;
````

**Part F: practice data for the next lessons.** Build the same kind of inline data the later labs use, and read it:

````python
%sql
CREATE OR REPLACE TEMP VIEW orders AS
SELECT * FROM VALUES
  (1, 'Asha', 120),
  (2, 'Ben',  75)
AS t(order_id, customer, amount);
````

````python
%sql
SELECT * FROM orders;
````
