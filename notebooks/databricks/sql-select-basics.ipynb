# SQL basics: SELECT, WHERE, ORDER BY and LIMIT

A query names the columns you want (SELECT), the table they come from (FROM), the rows to keep (WHERE), the order (ORDER BY) and how many rows to return (LIMIT). Knowing the order in which these clauses are evaluated explains most beginner errors.

**How to run:** attach **Serverless** compute (top right), then run the cells from top to bottom. Each part asks you to predict the result first; the expected results are listed at the end.

From the Learning Hub lesson: [SQL basics: SELECT, WHERE, ORDER BY and LIMIT](https://www.datapreplab.com/hub#topic/sql-select-basics) · Independent study resource, not affiliated with or endorsed by Databricks.

**Goal:** run the five clauses and predict each result before running it.

**Part A: build the data.**

````python
%sql
CREATE OR REPLACE TEMP VIEW orders AS
SELECT * FROM VALUES
  (1, 'Asha', 'UK', 120),
  (2, 'Ben',  'UK', NULL),
  (3, 'Chen', 'FR', 300),
  (4, 'Dee',  'UK', 75),
  (5, 'Eli',  'UK', 210),
  (6, 'Fay',  'UK', 75)
AS t(order_id, customer, country, amount);
````

**Part B: filter and sort.** Predict the order of the rows, then run:

````python
%sql
SELECT order_id, customer, amount
FROM orders
WHERE country = 'UK'
ORDER BY amount;
````

Ben's NULL amount comes **first**. Add `NULLS LAST` and run again.

**Part C: the NULL trap.** Predict how many rows each query returns:

````python
%sql
SELECT count(*) FROM orders WHERE amount = NULL;     -- 0
SELECT count(*) FROM orders WHERE amount IS NULL;    -- 1
````

**Part D: top three, deterministic.**

````python
%sql
SELECT order_id, customer, amount
FROM orders
WHERE country = 'UK' AND amount IS NOT NULL
ORDER BY amount DESC, order_id
LIMIT 3;
````

Expected evidence:

```text
Part B -> 2 Ben NULL, 4 Dee 75, 6 Fay 75, 1 Asha 120, 5 Eli 210 (4 and 6 may swap)
Part C -> 0 and 1
Part D -> 5 Eli 210, 1 Asha 120, 4 Dee 75
```
