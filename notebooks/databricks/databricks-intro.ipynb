# What is Databricks? The platform, your data and a first tour

Databricks is a platform for data, analytics and AI that runs in the cloud. It does not replace your storage; it connects to data in cloud object storage (your own, or storage Databricks manages for you) and gives you one place to store tables, govern them, process them with SQL, Python and Spark, and schedule the work. Free Edition lets you learn on it at no cost.

**How to run:** attach **Serverless** compute (top right), then run the cells from top to bottom. Each part asks you to predict the result first; the expected results are listed at the end.

**Goal:** create your free workspace, find your way around, and run your first query.

**Part A: create your workspace.** Go to the Free Edition sign-up page (<https://login.databricks.com/?intent=CE_SIGN_UP>), continue with Google or Microsoft or enter your email and the code Databricks sends you. Your workspace opens when it is ready.

**Part B: the tour.** Click through the sidebar and find:

1. **Workspace**: your home folder, where imported notebooks go.
2. **Catalog**: open the `workspace` catalog and its `default` schema. It is empty for now; your lab tables will appear here.
3. **SQL Editor**: where you will run the SQL labs.
4. **Compute**: serverless compute is already available; you do not need to create a cluster.

**Part C: your first query.** Open the **SQL Editor**, make sure a serverless SQL warehouse is selected at the top, and run:

````python
%sql
SELECT 'Hello, Databricks' AS greeting, current_date() AS today;
````

````python
%sql
SELECT current_catalog() AS catalog, current_schema() AS schema;
````
