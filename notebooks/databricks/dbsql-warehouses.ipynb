# Databricks SQL and SQL warehouses

Databricks SQL is the analyst's workspace in Databricks, the SQL editor, query history, dashboards, alerts and Genie, and it runs on SQL warehouses. A SQL warehouse is compute for SQL only. Serverless is the recommended type, starts in seconds and is the only type with Intelligent Workload Management; pro and classic run in your cloud account and take minutes to start. Size sets how fast one query runs; the number of clusters sets how many queries run at once.

**How to run:** attach **Serverless** compute (top right), then run the cells from top to bottom. Each part asks you to predict the result first; the expected results are listed at the end.

From the Learning Hub lesson: [Databricks SQL and SQL warehouses](https://www.datapreplab.com/hub#topic/dbsql-warehouses) · Independent study resource, not affiliated with or endorsed by Databricks.

**Goal:** find your warehouse, run a query on it and use a named parameter.

**Part A: find your compute.** In the sidebar open **SQL Warehouses**. Note the warehouse type, size and auto stop. On Free Edition it is the **Serverless Starter Warehouse**: type Serverless, size 2X-Small. Then run:

````python
%sql
SELECT current_catalog() AS catalog, current_schema() AS schema, current_user() AS me;
````

**Part B: a small sales table.**

````python
%sql
CREATE OR REPLACE TABLE workspace.default.sales_lab AS
SELECT * FROM VALUES
  (1, 'EU', DATE'2026-10-01', 120.00), (2, 'EU', DATE'2026-10-02', 80.00),
  (3, 'US', DATE'2026-10-01', 200.00), (4, 'US', DATE'2026-10-03', 50.00),
  (5, 'APAC', DATE'2026-10-02', 90.00)
AS t(order_id, region, order_date, amount);
````

**Part C: revenue by region.** Predict the EU total, then run:

````python
%sql
SELECT region, count(*) AS orders, sum(amount) AS revenue
FROM workspace.default.sales_lab
GROUP BY region ORDER BY revenue DESC;
````

**Part D: a parameter (SQL editor).** Open a new query in the **SQL editor**, select your SQL warehouse and paste:

```text
SELECT region, sum(amount) AS revenue
FROM workspace.default.sales_lab
WHERE region = :region
GROUP BY region;
```

Enter `EU` for `region` in the **Parameters** side panel (turn it on with **Parameters panel: ON** if it is hidden) and run it, then try `US`. Open **Query History** and find both runs.

Expected evidence:

```text
Part C -> US 2 orders 250.00 ; EU 2 orders 200.00 ; APAC 1 order 90.00
Part A -> Serverless Starter Warehouse, Serverless, 2X-Small
Part D -> EU 200.00 ; US 250.00 ; both runs listed in Query History
```

**Cleanup:** `DROP TABLE workspace.default.sales_lab;`
