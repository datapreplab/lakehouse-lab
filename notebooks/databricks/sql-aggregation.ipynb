# Aggregation: GROUP BY, HAVING and counting correctly

Aggregate functions such as count, sum and avg collapse many rows into one value. GROUP BY produces one result row per group, WHERE filters rows before grouping and HAVING filters groups after. count(*) counts rows; count(column) skips NULLs.

**How to run:** attach **Serverless** compute (top right), then run the cells from top to bottom. Each part asks you to predict the result first; the expected results are listed at the end.

From the Learning Hub lesson: [Aggregation: GROUP BY, HAVING and counting correctly](https://www.datapreplab.com/hub#topic/sql-aggregation) · Independent study resource, not affiliated with or endorsed by Databricks.

**Goal:** predict and compare count(*), count(column), sum and avg, then filter groups.

**Part A: data.**

````python
%sql
CREATE OR REPLACE TEMP VIEW emp AS
SELECT * FROM VALUES
  ('Asha', 'Data', 'a@x', 60000),
  ('Ben',  'Data', NULL,  40000),
  ('Chen', 'Data', 'c@x', NULL),
  ('Dee',  'Ops',  'd@x', 45000),
  ('Eli',  'Ops',  'e@x', 55000),
  ('Fay',  'HR',   'f@x', 50000)
AS t(name, dept, email, salary);
````

**Part B: whole-table counts.** Predict, then run:

````python
%sql
SELECT count(*), count(email), count(salary), avg(salary) FROM emp;
-- 6, 5, 5, 50000
````

**Part C: per department, with HAVING.**

````python
%sql
SELECT dept, count(*) AS employees, count(email) AS with_email, avg(salary) AS avg_salary
FROM emp
GROUP BY dept
HAVING count(*) >= 2
ORDER BY dept;
````

**Part D: break the rule on purpose.** In a new cell, run `SELECT dept, name, count(*) FROM emp GROUP BY dept` and read the error. Then fix it with `GROUP BY ALL`:

````python
%sql
SELECT dept, name, count(*) AS n FROM emp GROUP BY ALL ORDER BY dept, name;
````
