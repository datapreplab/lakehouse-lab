# Python in Databricks notebooks: cells, SQL and handing off to Spark

A Databricks notebook mixes Python, SQL and Markdown cells that share one session. A SparkSession called spark is ready to use, spark.sql runs SQL from Python and returns a DataFrame, and display shows results as a table or chart. Plain Python runs on one machine, so large data should stay in Spark DataFrames rather than be pulled into Python lists or pandas.

**How to run:** attach **Serverless** compute (top right), then run the cells from top to bottom. Each part asks you to predict the result first; the expected results are listed at the end.

From the Learning Hub lesson: [Python in Databricks notebooks: cells, SQL and handing off to Spark](https://www.datapreplab.com/hub#topic/py-databricks-notebooks) · Independent study resource, not affiliated with or endorsed by Databricks.

**Goal:** move between Python and SQL in one notebook and keep heavy work in Spark.

**Part A: Python to Spark.**

````python
orders = spark.createDataFrame(
    [(1, "UK", 120), (2, "FR", 80), (3, "UK", 50), (4, "DE", 90)],
    ["order_id", "region", "amount"])
orders.createOrReplaceTempView("orders")
display(orders)
````

**Part B: SQL on the same data.** In a new cell:

````python
%sql
SELECT region, count(*) AS orders FROM orders GROUP BY region
````

**Part C: back to Python.** Aggregate in Spark, then bring only the one-row answer back:

````python
summary = spark.sql("""
    SELECT region, sum(amount) AS revenue
    FROM orders GROUP BY region ORDER BY revenue DESC""")
display(summary)
````

````python
top = summary.first()
print(f"{top['region']} has the highest revenue: {top['revenue']}")
````
