# NULLs and CASE: missing values and conditional logic

NULL means unknown, and most expressions with NULL return NULL. coalesce picks the first non-NULL value, nullif turns a value into NULL, and CASE WHEN chooses a value by condition. Together they clean data and build categories.

**How to run:** attach **Serverless** compute (top right), then run the cells from top to bottom. Each part asks you to predict the result first; the expected results are listed at the end.

From the Learning Hub lesson: [NULLs and CASE: missing values and conditional logic](https://www.datapreplab.com/hub#topic/sql-nulls-case) · Independent study resource, not affiliated with or endorsed by Databricks.

**Goal:** see NULLs spread, fix them deliberately and build buckets.

**Part A: data.**

````python
%sql
CREATE OR REPLACE TEMP VIEW orders AS
SELECT * FROM VALUES
  (1, 'Asha', 'Leeds', 250, 20),
  (2, 'Ben',  NULL,     90, NULL),
  (3, 'Chen', 'York',  120, 0)
AS t(id, customer, city, price, discount);
````

**Part B: NULL spreads.** Predict Ben's value in each column:

````python
%sql
SELECT id, price - discount AS net, discount > 0 AS has_discount, city = NULL AS bad_test
FROM orders;
````

All three are NULL for Ben, and `bad_test` is NULL for everyone.

**Part C: clean and bucket.**

````python
%sql
SELECT id, customer,
       coalesce(city, 'unknown') AS city,
       price - coalesce(discount, 0) AS net_price,
       CASE WHEN price >= 200 THEN 'large'
            WHEN price >= 100 THEN 'medium'
            ELSE 'small' END AS size
FROM orders ORDER BY id;
````

**Part D: safe division.**

````python
%sql
SELECT 10 / nullif(0, 0) AS safe, try_divide(10, 0) AS also_safe;   -- NULL, NULL
````

Expected evidence:

```text
Part B -> Ben: NULL, NULL, NULL ; bad_test NULL for all rows
Part C -> 1 Asha Leeds 230 large ; 2 Ben unknown 90 small ; 3 Chen York 120 medium
Part D -> NULL | NULL
```
